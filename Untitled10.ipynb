```python
#decorators wrp a function to add behavior before or after it
def timer_decorator(func):
    import time
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print(f"{func.__name__} ran in{end -start:.4f}s")
    return wrapper
@timer_decorator
def compute_sum(n):
    return sum(range(n))
print(compute_sum(1000000))
```

```python
#fuction with exception handling
def safe_divide(a, b):
    try:
         return a / b
    except ZeroDivisionError:
        return "Error: Division by zero"
    except TypeError:
        return "Error: Invalid types"
print(safe_divide(10, 2))
print(safe_divide(10, 0))
print(safe_divide(10,  "x"))
```

```python
#generation function
def prime_generator(limit):
    for num in range(2, limit + 1):
        is_prime = all(num % i != 0 for i in range(2, int(num**0.5) + 1))
        if is_prime:
               yield num
primes = list(prime_generator(30))
print("Primes up to 30:", primes)
```

```python
# memoization using functools cache
from functools import lru_cache
@lru_cache(maxsize=None)
def fib(n):
    if n <= 1:
      return n
    return fib(n - 1) + fib(n - 2)
for i in [10,20,30,40]:
    print(f"fib({i}) = {fib(i)}")
```
